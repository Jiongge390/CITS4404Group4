# 03 — Algorithm Comparison

Run all implemented algorithms on training data with equal eval budget. Compare convergence curves.

```python
import sys; sys.path.insert(0, '../..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from part2.bot.bot import Bot
from part2.algorithms.random_search import RandomSearch
# from part2.algorithms.algorithm1 import Algorithm1
# from part2.algorithms.algorithm2 import Algorithm2

train = pd.read_csv('../../data/processed/train.csv', parse_dates=['date'])
P = train['close'].values
bot = Bot(P)

DIM = Bot.PARAM_DIM
BOUNDS = [(0, 1)] * 3 + [(2, 60)] * 3 + [(0.01, 0.99)] + \
         [(0, 1)] * 3 + [(2, 60)] * 3 + [(0.01, 0.99)]
MAX_EVALS = 5000
```

```python
results = {}

rs = RandomSearch(bot.evaluate, DIM, BOUNDS, max_evals=MAX_EVALS, seed=42)
bp, bf, hist = rs.optimise()
results['RandomSearch'] = {'params': bp, 'fitness': bf, 'history': hist}
print(f'RandomSearch best fitness: {bf:.2f}')

# Add other algorithms here once implemented
```

```python
# Convergence plot
plt.figure(figsize=(10, 4))
for name, r in results.items():
    evals, fitness = zip(*r['history'])
    plt.plot(evals, fitness, label=name)
plt.xlabel('Evaluations'); plt.ylabel('Best Fitness ($)')
plt.title('Algorithm Convergence'); plt.legend(); plt.tight_layout()
plt.savefig('../results/figures/convergence.png', dpi=150)
plt.show()
```
