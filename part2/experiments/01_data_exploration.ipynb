# 01 — Data Exploration

Load the Kaggle Bitcoin dataset, split into train/test, and visualise OHLCV data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load daily data
df = pd.read_csv('../../data/raw/BTC-Daily.csv', parse_dates=['date'])
df = df.sort_values('date').reset_index(drop=True)
print(df.shape)
df.head()
```

```python
# Train / test split (pre-2020 = train, 2020+ = test)
train = df[df['date'] < '2020-01-01'].copy()
test  = df[df['date'] >= '2020-01-01'].copy()
print(f'Train: {len(train)} rows, Test: {len(test)} rows')
```

```python
# Save preprocessed splits
import os
os.makedirs('../../data/processed', exist_ok=True)
train.to_csv('../../data/processed/train.csv', index=False)
test.to_csv('../../data/processed/test.csv', index=False)
print('Saved.')
```
