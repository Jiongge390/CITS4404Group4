# 02 — Bot Building Blocks

Verify filters and bot signals on training data. Reproduce figures from the spec.

```python
import sys; sys.path.insert(0, '../..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from part2.bot.filters import wma, sma_filter, lma_filter, ema_filter
from part2.bot.bot import Bot
from part2.bot.evaluator import backtest

train = pd.read_csv('../../data/processed/train.csv', parse_dates=['date'])
P = train['close'].values
```

```python
# Plot SMA 10 vs SMA 40
sma10 = wma(P, 10, sma_filter(10))
sma40 = wma(P, 40, sma_filter(40))

plt.figure(figsize=(14, 4))
plt.plot(P[-500:], label='Price', alpha=0.5)
plt.plot(sma10[-500:], label='SMA 10')
plt.plot(sma40[-500:], label='SMA 40')
plt.legend(); plt.title('SMA Filters'); plt.tight_layout(); plt.show()
```

```python
# Quick sanity check: default bot on training data
bot = Bot(P)
# Example: equal SMA/LMA/EMA weights, 10/20/15-day windows
params = [1, 1, 1, 10, 10, 10, 0.3,
          1, 1, 1, 40, 40, 40, 0.3]
print('Fitness:', bot.evaluate(params))
```
